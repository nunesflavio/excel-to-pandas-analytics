```python
import pandas as pd

# 1. Carregar a base de dados detalhada
caminho_arquivo = "../data/raw/loja_informatica_2026.xlsx"
df = pd.read_excel(caminho_arquivo, sheet_name="Dados Detalhados")

# 2. Conversão segura da coluna de Data
# O parâmetro format='%d/%m/%Y' diz exatamente como a data está escrita na planilha brasileira
df['Data'] = pd.to_datetime(df['Data'], format='%d/%m/%Y')

# Verificando se a conversão deu certo (agora o Dtype será datetime64)
print("Tipo da coluna Data após conversão:", df['Data'].dtype)


# 3. Recálculo seguro das colunas (substituindo as fórmulas vazias do Excel)
# Faturamento Total = Preço Unitário * Quantidade Vendida
df['Faturamento Total (R$)'] = df['Preço Unitário (R$)'] * df['Qtd Vendida']


# 4. Criando uma lógica limpa para o Status do Estoque via Python
# Regra: Estoque < 10 = 'Crítico', < 25 = 'Baixo', caso contrário = 'Normal'
def definir_status_estoque(estoque):
    if estoque < 10:
        return 'Crítico'
    elif estoque < 25:
        return 'Baixo'
    else:
        return 'Normal'

# Aplicando a função na coluna de Estoque Atual
df['Status Estoque'] = df['Estoque Atual'].apply(definir_status_estoque)


# 5. Visualizando as primeiras linhas com os dados limpos e recalculados
print("\n--- Amostra Tratada ---")
print(df[['ID Transação', 'Data', 'Produto', 'Preço Unitário (R$)', 'Qtd Vendida', 'Faturamento Total (R$)', 'Status Estoque']].head())

# 6. Salvando o resultado tratado em um novo arquivo Excel na pasta processed
df.to_excel("../data/processed/loja_informatica_tratada.xlsx", index=False)
print("\nArquivo tratado salvo com sucesso em data/processed/!")
```
